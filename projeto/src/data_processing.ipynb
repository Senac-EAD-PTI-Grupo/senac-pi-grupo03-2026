```python
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR = Path.cwd().parent / "data"
INPUT_FILE = DATA_DIR / "student_dropout_dataset_v3.csv"
OUTPUT_FILE = DATA_DIR / "base_tratada_v2.csv"

df = pd.read_csv(INPUT_FILE)
```

```python
# Total de alunos que desistiram
df.Dropout.sum()
```

Para evitar enviesar os dados, preenchendo com mediana, estou propondo a remoção de linhas com valores nulos e logo verificando se foram retirados muitos dados de alunos que saíram (Métrica alvo).
No total foram removidas 236 linhas de estudantes que abandonaram seus cursos.

Outra possibilidade preencher os valores medianos tirando como base outra coluna, por exemplo, no caso da escolaridade, baseado nos alunos que **saíram**, preencher com a moda.

```python
df.isnull().sum()
```

```python
df_sem_nulos = df.dropna()
```

```python
# Verificando quantos alunos foram removidos que saíram = 236
df_sem_nulos.Dropout.sum()
```

```python
# Verificando se sobraram valores nulos
df_sem_nulos.isnull().sum()
```

```python
len(df_sem_nulos)
```

```python
# Verificando se apenas tem Yes ou No
df_sem_nulos['Internet_Access'].unique()
```

```python
# Forma de contar valores e mostra caso tenha null
df['Internet_Access'].value_counts(dropna=False)
```

```python
mapeamento_yes_no = {'Yes': 1, 'No': 0}
df_sem_nulos['Internet_Access'] = df_sem_nulos['Internet_Access'].map(mapeamento_yes_no) # convertendo a coluna de acesso a internet
df_sem_nulos['Part_Time_Job'] = df_sem_nulos['Part_Time_Job'].map(mapeamento_yes_no) # convertendo a coluna de trabalho meio periodo
df_sem_nulos['Scholarship'] = df_sem_nulos['Scholarship'].map(mapeamento_yes_no) # convertendo a coluna de bolsa de estudos
```

```python
df_sem_nulos = pd.get_dummies(df_sem_nulos, columns=['Department'])
```

```python
df_sem_nulos.head()
```

```python
df_sem_nulos.columns
```

```python
df_sem_nulos.isnull().sum()
```

```python
df_sem_nulos.Parental_Education.unique()
```

```python
mapeamento_educacao = {
    'High School': 0,
    'Bachelor': 1,
    'Master': 2,
    'PhD': 3
}

df_sem_nulos['Parental_Education'] = df_sem_nulos['Parental_Education'].map(mapeamento_educacao)
```

```python
df_sem_nulos.isnull().sum()
```

```python
df_sem_nulos.to_csv(OUTPUT_FILE, index=False)
```
